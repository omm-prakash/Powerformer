```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
import re

data_path = '/scratch/opsahoo.eee20.itbhu/Thesis/SmartGrid Datasets'
```

```python
def extract_case_range(path):
    pattern = re.compile(r'ICaseNum_rms_(\d+)Results\.csv$')
    file_names = os.listdir(path)
    min_case, max_case = float('inf'), 0

    for name in file_names:
        match = pattern.search(name)
        if match:
            case = int(match.group(1))
            min_case = min(min_case, case)
            max_case = max(max_case, case)

    return min_case, max_case
```

```python
def plot_random_case(data_path, data_name):
    path = os.path.join(data_path, data_name)
    min_case, max_case = extract_case_range(path)
    random_case_num = np.random.randint(min_case, max_case+1)

    print(f'Data Type: {data_name}\nRandom Case: {random_case_num}')

    current_file_format = 'ICaseNum_rms_{}Results.csv'
    voltage_file_format = 'VCaseNum_rms_{}Results.csv'

    current_df = pd.read_csv(os.path.join(path, current_file_format.format(random_case_num)), index_col=False)
    voltage_df = pd.read_csv(os.path.join(path, voltage_file_format.format(random_case_num)), index_col=False)

    plt.figure(figsize=(15,5))
    plt.plot(current_df[['I121A_rms[kA]', 'I121B_rms[kA]', 'I121C_rms[kA]',
    'I232A_rms[kA]', 'I232B_rms[kA]', 'I232C_rms[kA]', 'I264A_rms[kA]',
    'I264B_rms[kA]', 'I264C_rms[kA]', 'I343A_rms[kA]', 'I343B_rms[kA]',
    'I343C_rms[kA]', 'I656A_rms[kA]', 'I656B_rms[kA]', 'I656C_rms[kA]',
    'I765A_rms[kA]', 'I765B_rms[kA]', 'I765C_rms[kA]', 'S3IA_rms[kA]',
    'S3IB_rms[kA]', 'S3IC_rms[kA]']])
    plt.ylabel('Current RMS (kA)')
    plt.xlabel('Time Instance')
    plt.title(f'Data Name: "{data_name}" | Case No. {random_case_num} | Fault Type: {current_df['FAULT_TYPE'].iloc[0].item()}')
    plt.show()

    plt.figure(figsize=(15,5))
    plt.plot(voltage_df[['N16_rms[kV]',
        'N17_rms[kV]', 'N18_rms[kV]', 'N19_rms[kV]', 'N1_rms[kV]',
        'N20_rms[kV]', 'N21_rms[kV]', 'N2_rms[kV]', 'N3_rms[kV]', 'N4_rms[kV]',
        'N5_rms[kV]', 'N6_rms[kV]', 'N7_rms[kV]', 'N8_rms[kV]', 'N9_rms[kV]']])
    plt.ylabel('Volatage RMS (kV)')
    plt.xlabel('Time Instance')
    plt.title(f'Data Name: "{data_name}" | Case No. {random_case_num} | Fault Type: {current_df['FAULT_TYPE'].iloc[0].item()}')
    plt.show()
```

```python
# data_name = 'microgrid_steady_state_rms'
data_name = 'microgrid_fault_dataset_rms'

plot_random_case(data_path, data_name)
```
