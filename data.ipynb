```python
import re

def get_max_expt_number(data):
    expt_numbers = []
    for entry in data:
        match = re.search(r'expt-(\d+)', entry)
        if match:
            expt_numbers.append(int(match.group(1)))
    return max(expt_numbers) if expt_numbers else 0

# Example data
data = [
    "expt-10| 2025-04-02 13:34", "expt-2| 2025-03-30 23:28", "expt-5| 2025-03-31 17:27", "expt-8| 2025-04-01 17:40",
    "expt-20| 2025-04-02 14:17", "expt-3| 2025-03-31 02:07", "expt-6| 2025-04-01 00:24", "expt-9| 2025-04-02 00:37",
    "expt-1| 2025-03-30 19:51", "expt-4| 2025-03-31 09:08", "expt-7| 2025-04-01 09:47"
]
data = []

max_expt = get_max_expt_number(data)
print("Maximum experiment number:", max_expt)
```
